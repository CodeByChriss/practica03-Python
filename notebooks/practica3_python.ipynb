# Ejercicio 1 — Catálogo de productos como lista

```python
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]

# Ejercicio 1.1
print(f"El tamaño del catálogo es de: {len(catalogo)}")
# Se divide entre 2 ya que el array usa un espacio para el nombre
# y otro para el precio por lo que un producto son 2 espacios
print(f"Hay un total de {len(catalogo)/2} productos.")

# Ejercicio 1.2
# [inicio : fin : paso]
nombres = catalogo[::2]
print(f"Solo nombres: {nombres}")
precios = catalogo[1::2]
print(f"Solo precios: {precios}")

# Ejercicio 1.3
print(f"El precio del Robot Aspirador es de {precios[nombres.index("Robot Aspirador")]}")

# Ejercicio 1.4
ultimos_tres_nombres = nombres[-3:]
print(ultimos_tres_nombres)

# Ejercicio 1.5
# Sumamos todos los precios y lo dividimos por su cantidad
precio_medio = round(sum(precios) / len(precios), 2)
print(f"El precio medio del catálogo es de: {precio_medio}")
```

# Ejercicio 2 — Inventario multialmacén con listas anidadas

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

# Ejercicio 2.1
# 4ª fila se traduce a 3 ya que python empieza por 0
print(f"El nombre del producto de la cuarta fila es de {inventario[3][1]}")

# Ejercicio 2.2
# Itero sobre inventario y creo directamente el array
valor_stock = [[inv[1], round(inv[2]*inv[3], 2)] for inv in inventario]
print(valor_stock)

# Ejercicio 2.3
# Sumo cada valor del stock
valor_inmovilizado = sum(valor[1] for valor in valor_stock)
print(f"El valor inmovilizado total es de: {valor_inmovilizado}")

# Ejercicio 2.4
inventario_alm_sur = inventario[2:4]
valor_alm_sur = sum(round(alm[3]*alm[2], 2) for alm in inventario_alm_sur)
print(f"El valor de ALM-SUR es de: {valor_alm_sur}")
```

# Ejercicio 3 — Manipulación destructiva y copias de listas

```python
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00]

# 3.1
precios_rebajados = precios_originales
precios_rebajados[0] = 49.90
print(precios_originales)
print(precios_rebajados)
# Lo que ocurre es que ambas listas apuntan al mismo sitio en memoria por lo que modificar una modifica la otra

# 3.2
precios_rebajados[0] = 59.90
list_precios = list(precios_originales)
precios_rebajados = precios_originales[:]
list_precios[0] = 29.90
precios_rebajados[0] = 39.90
print(list_precios)
print(precios_rebajados)
print(precios_originales)

# 3.3
for i in range(len(precios_rebajados)):
    precios_rebajados[i] = round(precios_rebajados[i] * 0.90, 2)
print(precios_rebajados)

# 3.4
precios_originales.extend([330.0])
precios_originales.extend([380.0])
precios_originales.insert(2, 444.44)
print(precios_originales)
del precios_originales[4]
descartado = precios_originales.pop()
print(precios_originales)
print(f"Descartado: {descartado}")

# 3.5
print(id(precios_originales))
print(id(precios_rebajados))
print(id(list_precios))
```
