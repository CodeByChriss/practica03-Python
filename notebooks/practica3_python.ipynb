# Ejercicio 1 — Catálogo de productos como lista

```python
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]

# Ejercicio 1.1
print(f"El tamaño del catálogo es de: {len(catalogo)}")
# Se divide entre 2 ya que el array usa un espacio para el nombre
# y otro para el precio por lo que un producto son 2 espacios
print(f"Hay un total de {len(catalogo)/2} productos.")

# Ejercicio 1.2
# [inicio : fin : paso]
nombres = catalogo[::2]
print(f"Solo nombres: {nombres}")
precios = catalogo[1::2]
print(f"Solo precios: {precios}")

# Ejercicio 1.3
print(f"El precio del Robot Aspirador es de {precios[nombres.index("Robot Aspirador")]}")

# Ejercicio 1.4
ultimos_tres_nombres = nombres[-3:]
print(ultimos_tres_nombres)

# Ejercicio 1.5
# Sumamos todos los precios y lo dividimos por su cantidad
precio_medio = round(sum(precios) / len(precios), 2)
print(f"El precio medio del catálogo es de: {precio_medio}")
```

# Ejercicio 5 — Funciones propias para el cálculo de importes

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """
    calcular_importe(unidades, precio_unitario, descuento_pct=0)
    Devuelve unidades * precio_descuento
    siendo precio_descuento el precio_unitario
    con el descuento_pct ya aplicado.
    """
    return (precio_unitario * (1 - descuento_pct / 100)) * unidades

def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """
    resumen_pedido(unidades, precio_unitario, descuento_pct=0)
    Se calcula el importe_bruto usando la función calcular_importe()
    pasandole como descuento_pct 0 y se repite lo mismo para
    importe_neto pero se pasa el descuento_pct correcto.
    Con el importe_bruto y el importe_neto se calcula el ahorro
    restandolos.
    Devuelve: (importe_bruto, ahorro, importe_neto)
    """
    importe_bruto = calcular_importe(unidades, precio_unitario)
    importe_neto = calcular_importe(unidades, precio_unitario, descuento_pct)
    ahorro = importe_bruto - importe_neto
    return (importe_bruto, ahorro, importe_neto)

resultado = resumen_pedido(unidades = 5, precio_unitario = 10.0, descuento_pct = 20.0)
importe_bruto = resultado[0]
ahorro = resultado[1]
importe_neto = resultado[2]
print(f"Resultado: {resultado}")
print(f"Importe bruto: {importe_bruto}")
print(f"Ahorro: {ahorro}")
print(f"Importe neto: {importe_neto}")

print(help(calcular_importe))
print(help(resumen_pedido))

pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]
importes = []

for pedido in pedidos:
    importes.append(calcular_importe(pedido[0], pedido[1], pedido[2]))

print(f"El importe mayor es: {max(importes)}")
print(f"El importe minimo es: {min(importes)}")
print(f"Total de importes: {len(importes)}")
print(f"Importes ordenados: {sorted(importes)}")
```

# Ejercicio 6 — Normalización de códigos SKU con métodos

```python
skus_crudos = [
    "  elc-0012-es ", "HOG-0045-ES", "dep-0003-pt  ",
    "ofi-0021-es", "  ELC-0012-ES", "hog-0099-fr ",
]

skus_limpios = [sku.strip().upper() for sku in skus_crudos]
print(skus_limpios)

veces = skus_limpios.count("ELC-0012-ES")
print(f"ELC-0012-ES aperece un total de {veces}")

print(f"La primera vez que aparece OFI-0021-ES es en el index {skus_limpios.index("OFI-0021-ES")}")

partes = [sku.split("-") for sku in skus_limpios]
print(partes)

skus_es = []
for sku in skus_limpios:
    # Hacemos split por el guión y nos quedamos con el país
    pais = sku.split("-")[2]
    if pais == 'ES':
        skus_es.append(sku.replace("-","."))
print(skus_es)

cadena = " | ".join(skus_limpios)
print(cadena)
```

# Ejercicio 11 — Estadística descriptiva sobre telemetría simulada

```python
rng = np.random.default_rng(2024)
temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

media_temp = temperatura.mean()
mediana_temp = np.median(temperatura)
print(f"Temperatura media: {media_temp}")
print(f"Mediana: {mediana_temp}")
print(f"Desviación típica: {temperatura.std()}")
print(f"Temperatura mínima: {min(temperatura)}")
print(f"Temperatura máxima: {max(temperatura)}")

percentiles = np.percentile(vibracion, [25, 50, 75, 95])
print("Percentiles vibración (25, 50, 75, 95):", percentiles)

matriz_telemetria = np.column_stack([temperatura, vibracion])
print("Shape matriz telemetría:", matriz_telemetria.shape)

correlacion = np.corrcoef(temperatura, vibracion)
print("Matriz de correlación:\n", correlacion)

umbral = media_temp + 2 * temperatura.std()
lecturas_criticas = (temperatura > umbral).sum()
porcentaje_criticas = (lecturas_criticas / len(temperatura)) * 100
print(f"Lecturas que superan el umbral: {lecturas_criticas} ({porcentaje_criticas}%)")
```
